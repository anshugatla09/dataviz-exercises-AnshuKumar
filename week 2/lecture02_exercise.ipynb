## Setup — Run this cell first


```python
import pandas as pd
import numpy as np

# World Happiness Report 2023 — representative data
# Source: https://www.kaggle.com/datasets/ajaypalsinghlo/world-happiness-report-2023

df = pd.read_csv('../data/world_happiness_2023.csv')
df.columns = ['Country','Region','Happiness_Score','GDP','Social_Support',
              'Life_Expectancy','Freedom','Generosity','Corruption']


print(f"Dataset: {len(df)} countries, {len(df.columns)} columns")
print(df.head())
```

```python
import plotly.express as px
import plotly.graph_objects as go

# Explore the dataset before you start
print("Regions in dataset:")
print(df['Region'].value_counts())
print("\nScore range:", df['Happiness_Score'].min(), "–", df['Happiness_Score'].max())
print("\nBottom 10 countries:")
print(df.nsmallest(10, 'Happiness_Score')[['Country','Region','Happiness_Score']])
```

## Task 1 — Regional Comparison Bar Chart

**What to build:** A horizontal bar chart showing the **average happiness score by region**, sorted from highest to lowest.

**Requirements:**
- Horizontal orientation (category names are long)
- Sorted by score, descending (so the happiest region is at the top)
- Zero baseline on x-axis
- At least one design choice that goes beyond the Plotly default (colour, annotation, labels, etc.)
- An insight title that answers: *which region stands out and why does it matter?*

**Hint:** Use `df.groupby('Region')['Happiness_Score'].mean()` to compute the averages.


```python
# Task 1: Regional comparison bar chart
# -------------------------------------

# Step 1: Compute average happiness score by region
region_avg = (df.groupby('Region')['Happiness_Score']
              .mean()
              .reset_index()
              .sort_values('Happiness_Score'))  # sort for horizontal bar

print(region_avg)

# Step 2: Build your chart
# YOUR CODE HERE
region_avg = (df.groupby('Region')['Happiness_Score']
              .mean()
              .reset_index()
              .sort_values('Happiness_Score'))

print(region_avg)

global_avg = df['Happiness_Score'].mean()
top_region = region_avg.iloc[-1]
bottom_region = region_avg.iloc[0]

colors = ['#C0392B' if r == bottom_region['Region'] else
          '#1B9E77' if r == top_region['Region'] else '#B0B7C3'
          for r in region_avg['Region']]

fig1 = go.Figure(go.Bar(
    x=region_avg['Happiness_Score'],
    y=region_avg['Region'],
    orientation='h',
    marker_color=colors,
    text=region_avg['Happiness_Score'].round(2),
    textposition='outside',
    cliponaxis=False,
    hovertemplate='%{y}: %{x:.2f}<extra></extra>'
))

fig1.add_vline(x=global_avg, line_dash='dash', line_color='#444',
               annotation_text=f'Global avg {global_avg:.2f}',
               annotation_position='bottom right')

gap = top_region['Happiness_Score'] - bottom_region['Happiness_Score']
fig1.update_layout(
    title=dict(text=(f"<b>{top_region['Region']} is the happiest region — "
                     f"{gap:.1f} points ahead of {bottom_region['Region']}</b>"),
               x=0.0, font=dict(size=20)),
    xaxis=dict(title='Average happiness score (0–10)', range=[0, 10],
               showgrid=True, gridcolor='#EEE', zeroline=True, zerolinecolor='#333'),
    yaxis=dict(title=''),
    plot_bgcolor='white', paper_bgcolor='white',
    font=dict(family='Arial', size=13),
    margin=dict(l=20, r=40, t=80, b=50),
    height=500, showlegend=False
)
fig1.show()
```

## Task 2 — Bottom vs. Top: A Contrast Story

**What to build:** A bar chart that highlights the **gap between the happiest and least happy countries**, focusing on a specific insight.

**Requirements:**
- Show the **top 8 AND bottom 8 countries** together (16 bars total)
- Use **colour** to distinguish the two groups (not Plotly's default rainbow)
- Add a **visual separator or annotation** that emphasises the gap
- Insight title that tells the story of the gap

**Hint:** Use `pd.concat([df.nlargest(8,'Happiness_Score'), df.nsmallest(8,'Happiness_Score')])` to get both groups.

**Stretch goal:** Add a vertical reference line showing the global average.


```python
# Task 2: Top 8 vs. Bottom 8 contrast
# ------------------------------------

# Step 1: Get top and bottom countries
top8 = df.nlargest(8, 'Happiness_Score').copy()
top8['Group'] = 'Top 8'
bottom8 = df.nsmallest(8, 'Happiness_Score').copy()
bottom8['Group'] = 'Bottom 8'

combined = pd.concat([bottom8, top8]).sort_values('Happiness_Score')
global_avg = df['Happiness_Score'].mean()
print(f"Global average: {global_avg:.2f}")

# Step 2: Build your chart
# YOUR CODE HERE
top8 = df.nlargest(8, 'Happiness_Score').copy()
top8['Group'] = 'Top 8'
bottom8 = df.nsmallest(8, 'Happiness_Score').copy()
bottom8['Group'] = 'Bottom 8'

combined = pd.concat([bottom8, top8]).sort_values('Happiness_Score')
global_avg = df['Happiness_Score'].mean()
print(f"Global average: {global_avg:.2f}")

color_map = {'Top 8': '#1B9E77', 'Bottom 8': '#C0392B'}
top_mean = top8['Happiness_Score'].mean()
bottom_mean = bottom8['Happiness_Score'].mean()

fig2 = go.Figure(go.Bar(
    x=combined['Happiness_Score'],
    y=combined['Country'],
    orientation='h',
    marker_color=combined['Group'].map(color_map),
    text=combined['Happiness_Score'].round(2),
    textposition='outside',
    cliponaxis=False,
    hovertemplate='%{y}: %{x:.2f}<extra></extra>'
))

# Global average line (stretch goal)
fig2.add_vline(x=global_avg, line_dash='dash', line_color='#444',
               annotation_text=f'Global avg {global_avg:.2f}',
               annotation_position='top')

# Separator between the two groups
fig2.add_hline(y=7.5, line_color='#888', line_width=1.5, line_dash='dot')

# Gap annotation
fig2.add_annotation(
    x=8.6, y=7.5, xref='x', yref='y',
    text=f'<b>{top_mean - bottom_mean:.1f}-point gap</b><br>(avg. of each group)',
    showarrow=False, font=dict(size=14, color='#333'),
    bgcolor='#F5F5F5', bordercolor='#888', borderwidth=1, borderpad=6
)

# Legend substitute
fig2.add_annotation(x=0.99, y=0.02, xref='paper', yref='paper', showarrow=False,
                    text='<span style="color:#1B9E77">■</span> Top 8   '
                         '<span style="color:#C0392B">■</span> Bottom 8',
                    font=dict(size=13), xanchor='right')

fig2.update_layout(
    title=dict(text=(f"<b>The happiest countries score nearly twice as high as the least happy "
                     f"({top_mean:.1f} vs {bottom_mean:.1f})</b>"),
               x=0.0, font=dict(size=20)),
    xaxis=dict(title='Happiness score (0–10)', range=[0, 10],
               showgrid=True, gridcolor='#EEE', zeroline=True, zerolinecolor='#333'),
    yaxis=dict(title=''),
    plot_bgcolor='white', paper_bgcolor='white',
    font=dict(family='Arial', size=13),
    margin=dict(l=20, r=40, t=90, b=50),
    height=650, showlegend=False
)
fig2.show()
```

## Done? Stretch Goal

If you finish both tasks with time to spare, try this:

**Task 3 (stretch):** Build a **grouped bar chart** comparing 2 sub-factors (e.g. `GDP_per_capita` and `Freedom`) across the 5 most populated regions. Use colour meaningfully and write an insight title.

Regions to include: `'Western Europe'`, `'Latin America'`, `'East Asia'`, `'Sub-Saharan Africa'`, `'South Asia'`


```python
# Stretch goal — grouped bar chart
# YOUR CODE HERE

regions = ['Western Europe', 'Latin America', 'East Asia', 'Sub-Saharan Africa', 'South Asia']
subset = df[df['Region'].isin(regions)]

# Fallback if region names in your CSV differ
if subset['Region'].nunique() < 5:
    print("Region names didn't match. Available:", df['Region'].unique())
    regions = df['Region'].value_counts().head(5).index.tolist()
    subset = df[df['Region'].isin(regions)]

factors = ['GDP', 'Freedom']
reg_means = subset.groupby('Region')[factors].mean().reset_index()
reg_means = reg_means.sort_values('GDP', ascending=False)
print(reg_means)

fig3 = go.Figure()
fig3.add_bar(x=reg_means['Region'], y=reg_means['GDP'], name='GDP per capita',
             marker_color='#2C6FBB', text=reg_means['GDP'].round(2), textposition='outside')
fig3.add_bar(x=reg_means['Region'], y=reg_means['Freedom'], name='Freedom',
             marker_color='#F39C12', text=reg_means['Freedom'].round(2), textposition='outside')

richest = reg_means.iloc[0]['Region']
poorest = reg_means.iloc[-1]['Region']
fig3.update_layout(
    barmode='group',
    title=dict(text=(f"<b>Wealth varies far more across regions than freedom does "
                     f"({richest} vs {poorest})</b>"),
               x=0.0, font=dict(size=20)),
    yaxis=dict(title='Average factor score', rangemode='tozero',
               showgrid=True, gridcolor='#EEE', zeroline=True, zerolinecolor='#333'),
    xaxis=dict(title=''),
    plot_bgcolor='white', paper_bgcolor='white',
    font=dict(family='Arial', size=13),
    legend=dict(orientation='h', y=1.08, x=0),
    margin=dict(l=20, r=40, t=110, b=50),
    height=520
)
fig3.show()
```
